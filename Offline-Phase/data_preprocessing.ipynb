Since we want hourly prediction the first thing we do is convert the timestamp into hourly timestamp

```python
df["timestamp"] = pd.to_datetime(df["timestamp"], utc=True)


df["timestamp"] = df["timestamp"].dt.floor("h")
```

Since the timestamp above is hourly now we try to aggregate all values that belong to the same timestamp and sensorId

```python
df_hourly = (
    df.groupby(["sensorId", "timestamp"])
      .agg({
          "city":"first",
          "humidity":"mean",
          "pm10":"mean",
          "pm25":"mean",
          "pressure":"mean",
          "temperature":"mean",
          "wind_speed":"mean"
      })
      .reset_index()
)
```

```python
df_hourly = df_hourly.sort_values(by=["timestamp", "sensorId"]).reset_index(drop=True)
```

Now we want to check which sensors have a lot of missing values and those that have over 100% missing data will be removed

```python
sensor_stats = df_hourly.groupby('sensorId')[['pm10', 'pm25']].count()
valid_sensors = sensor_stats[sensor_stats.sum(axis=1) > 0].index

df_hourly = df_hourly[df_hourly['sensorId'].isin(valid_sensors)]
```

Since there is still missing values across the dataset, now we try to interpolate the values of the humidity, temperature,pressure,wind_speed, pm10 and pm25 in order to reduce the missing value percentage based on the sensor id

```python
df_hourly = df_hourly.sort_values(['sensorId', 'timestamp'])

cols_to_fix = ['pm10', 'pm25', 'temperature', 'humidity', 'pressure', 'wind_speed']

for col in cols_to_fix:

    df_hourly[col] = df_hourly.groupby('sensorId')[col].transform(
        lambda x: x.interpolate(method='linear', limit=3)
    )
```

```python
percent_missing = (df_hourly.isnull().sum() * 100) / len(df_hourly)
print(percent_missing)
```

```python
df_hourly = df_hourly.dropna(subset=['pm10', 'pm25'])

print(f"Remaining rows after dropping missing targets: {len(df_hourly)}")
```

```python
percent_missing = (df_hourly.isnull().sum() * 100) / len(df_hourly)
print(percent_missing)
```

Fill missing weather with the average of all other sensors in Bitola for that specific hour

```python
weather_cols = ['humidity', 'pressure', 'temperature', 'wind_speed']


df_hourly[weather_cols] = df_hourly[weather_cols].fillna(
    df_hourly.groupby('timestamp')[weather_cols].transform('mean')
)
```

```python
percent_missing = (df_hourly.isnull().sum() * 100) / len(df_hourly)
print(percent_missing)
```

Since we have small percent of the missing values we use for example the ffil method

```python
df_hourly[weather_cols] = df_hourly[weather_cols].fillna(method='ffill')

# Verify one last time - everything should be 0.0
print(df_hourly.isnull().sum())
```

Function for appending neighborhood features

```python
def append_neighbors(df_hourly, neighbors_df, weather_cols, k_search=20, k_keep=3):
    # 1. Standardize the neighbor list
    # Ensure we only take the top K based on distance
    neighbors_topk = (
        neighbors_df.sort_values(["sensor_id", "distance_km"])
        .groupby("sensor_id")
        .head(k_search)
        .copy()
    )

    # Track original distance rank
    neighbors_topk['dist_rank'] = neighbors_topk.groupby("sensor_id").cumcount() + 1

    # 2. Merge with main data
    # We use 'neighbor_id' from the matrix to match 'sensorId' in the hourly data
    neighbor_values = neighbors_topk.merge(
        df_hourly[['sensorId', 'timestamp'] + weather_cols],
        left_on='neighbor_id',
        right_on='sensorId',
        how='inner'
    )

    # 3. Filter for availability
    # The 'sensor_id' here is the ORIGINAL sensor we are finding neighbors for
    available_topk = (
        neighbor_values.sort_values(['sensor_id', 'timestamp', 'dist_rank'])
        .groupby(['sensor_id', 'timestamp'])
        .head(k_keep)
        .copy()
    )

    # Create the 1, 2, 3 rank for the wide-format columns
    available_topk['final_rank'] = available_topk.groupby(['sensor_id', 'timestamp']).cumcount() + 1

    # 4. Pivot to wide format
    pivot_df = available_topk.pivot(
        index=['sensor_id', 'timestamp'],
        columns='final_rank',
        values=weather_cols
    )

    # Clean up column names: neighbor1_temp, neighbor2_temp, etc.
    if isinstance(pivot_df.columns, pd.MultiIndex):
        pivot_df.columns = [f"neighbor{rank}_{col}" for col, rank in pivot_df.columns]
    else:
        # Handle case with only one weather column
        pivot_df.columns = [f"neighbor{i}_{weather_cols[0]}" for i in pivot_df.columns]

    pivot_df = pivot_df.reset_index()

    # 5. Final Join back to original data
    df_result = df_hourly.merge(
        pivot_df,
        left_on=['sensorId', 'timestamp'],
        right_on=['sensor_id', 'timestamp'],
        how='left'
    ).drop(columns=['sensor_id'])

    return df_result
```

```python
features = ['temperature', 'humidity', 'pressure', 'wind_speed']

for feat in features:
    # 1. Fill Neighbor 3 with Neighbor 2 if 3 is missing
    n3_col = f'neighbor3_{feat}'
    n2_col = f'neighbor2_{feat}'
    n1_col = f'neighbor1_{feat}'

    df_final[n3_col] = df_final[n3_col].fillna(df_final[n2_col])

    # 2. Fill Neighbor 2 with Neighbor 1 if 2 is still missing
    df_final[n2_col] = df_final[n2_col].fillna(df_final[n1_col])

    # 3. Final safety check: if Neighbor 1 is missing (very rare),
    # fill with the sensor's own current value for that feature.
    df_final[n1_col] = df_final[n1_col].fillna(df_final[feat])
```

Function for extracting some important details about the timestamp

```python
def extract_time_features(df, timestamp_col='timestamp'):


    # 1. Cyclic Hour (24-hour cycle) //
    df['hour_sin'] = np.sin(2 * np.pi * df[timestamp_col].dt.hour / 24)
    df['hour_cos'] = np.cos(2 * np.pi * df[timestamp_col].dt.hour / 24)

    # 2. Cyclic Month (12-month cycle)//
    df['month_sin'] = np.sin(2 * np.pi * (df[timestamp_col].dt.month - 1) / 12)
    df['month_cos'] = np.cos(2 * np.pi * (df[timestamp_col].dt.month - 1) / 12)

    # 3. Weekend Flag
    # Monday=0, Sunday=6. Weekends are 5 and 6.
    df['is_weekend'] = df[timestamp_col].dt.dayofweek.isin([5, 6]).astype(int)

    # 4. Heating Season Flag (Specific for Bitola/Macedonia)
    # Most wood burning occurs from November (11) to March (3)
    df['is_heating_season'] = df[timestamp_col].dt.month.isin([11, 12, 1, 2, 3]).astype(int)

    return df
```
