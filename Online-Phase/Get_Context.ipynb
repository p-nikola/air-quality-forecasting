# Construct the context window

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/training/bitola_final_data.csv")
```

```python
df.describe()
```

```python
df.head(10)
```

```python
PREDICTION_LENGTH = 72
ID_COL = 'sensorId'
```

```python
ground_truth_df = df.groupby(ID_COL).apply(lambda x: x.iloc[-PREDICTION_LENGTH:]).reset_index(drop=True)
```

```python
ground_truth_df
```

```python
ground_truth_df.describe()
```

```python
ground_truth_df.isnull().sum()
```

```python
ground_truth_df.to_csv("context_bitola.csv", index=False)
```
